```python
import pandas as pd
import numpy as np
from pathlib import Path
from scipy.io import arff

data_dir = Path("../data")
arff_files = sorted(data_dir.glob("*.arff"))
print("File ARFF ditemukan:", [f.name for f in arff_files])

for file in arff_files:
    data, meta = arff.loadarff(file)
    df = pd.DataFrame(data)
    
    for col in df.select_dtypes([object]).columns:
        df[col] = df[col].apply(lambda x: x.decode("utf-8") if isinstance(x, bytes) else x)
    
    print(f"\n=== AUDIT: {file.name} ===")
    print(f"Shape        : {df.shape[0]} baris x {df.shape[1]} kolom")
    print(f"Missing vals : {df.isna().sum().sum()}")
    print(f"Duplikat     : {df.duplicated().sum()}")
    
    target_col = df.columns[-1]
    print(f"Target ({target_col}): {df[target_col].value_counts(dropna=False).to_dict()}")
    
    loc_cols = [c for c in df.columns if "loc" in c.lower()]
    if loc_cols:
        loc_col = loc_cols[0]
        zero_loc = (pd.to_numeric(df[loc_col], errors="coerce") <= 0).sum()
        print(f"Kolom LOC ({loc_col}) <= 0 : {zero_loc} baris")
```

```python
for fname in ["JM1.arff", "PC4.arff", "MC1.arff"]:
    data, _ = arff.loadarff(data_dir / fname)
    df = pd.DataFrame(data)
    loc_cols = [c for c in df.columns if "loc" in c.lower()]
    print(f"{fname} -> kolom LOC: {loc_cols}")
```
