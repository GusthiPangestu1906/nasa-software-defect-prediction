Notebook ini melakukan audit menyeluruh terhadap seluruh file dataset di folder `data/`:

- Nama kolom target defect dan kolom LOC (`LOC_TOTAL`)

- Catatan ketidakcocokan skema dibanding dataset NASA cleaned standar
- Visualisasi distribusi kelas (`../figures/class_distribution.png`)

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(str(Path("..").resolve()))
from src.audit_data import audit_directory

data_dir = Path("../data")
figures_dir = Path("../figures")
figures_dir.mkdir(parents=True, exist_ok=True)

summary_df = audit_directory(data_dir)
print(summary_df.to_string(index=False))
```

## Catatan Ketidakcocokan Skema (Schema Discrepancies)
1. **Format File**: Seluruh file dari Kaggle (`aczy156/nasa-software-defect-prediction`) berformat `.arff` (WEKA), bukan `.csv`.
2. **Nama Kolom Target**: Pada `JM1.arff`, kolom target bernama `label` (`N`/`Y`), sedangkan pada dataset lain bernama `Defective` (`N`/`Y`). Dataset PROMISE/NASA standar umumnya menggunakan `defects` (`false`/`true` atau `0`/`1`).
3. **Dataset Kosong**: `KC4.arff` merupakan file stub kosong (`0 baris x 1 kolom`).
4. **Variasi Jumlah Fitur**: `JM1` dan `KC1` memiliki 22 kolom (21 fitur + 1 target), sementara varian lain memiliki 37 hingga 40 kolom.
5. **Nilai `LOC_TOTAL <= 0`**: Tidak ditemukan pada `JM1`, `KC1`, `CM1`, `KC3`, `MC2`, `MW1`, namun ditemukan pada `MC1` (131 baris), `PC2` (11 baris), `PC4` (9 baris), `PC5` (20 baris), `PC1` (2 baris), `PC3` (2 baris).

```python
valid_df = summary_df[summary_df["Baris"] > 0].copy()
valid_df["Dataset"] = valid_df["File"].str.replace(".arff", "", regex=False).str.replace(".csv", "", regex=False)

fig, ax = plt.subplots(figsize=(11, 5))
df_plot = valid_df.melt(
    id_vars=["Dataset", "Defect_Pct"],
    value_vars=["NonDefective_N", "Defective_N"],
    var_name="Class",
    value_name="Count"
)
df_plot["Class"] = df_plot["Class"].map({"NonDefective_N": "Non-Defective (N)", "Defective_N": "Defective (Y)"})

sns.barplot(data=df_plot, x="Dataset", y="Count", hue="Class", ax=ax, palette=["#4c72b0", "#dd8452"])
ax.set_title("Distribusi Kelas (Class Imbalance) pada Dataset NASA MDP")
ax.set_ylabel("Jumlah Modul")
ax.set_xlabel("Varian Dataset")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig(figures_dir / "class_distribution.png", dpi=300)
plt.close(fig)
```
